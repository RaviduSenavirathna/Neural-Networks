Dataset (AND gate)

```python
import numpy as np

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [0],
    [0],
    [1]
])
```

Initialize weights and bias

```python
np.random.seed(42)

W = np.random.randn(2, 1)
b = np.zeros((1, 1))
```

Activation (sigmoid)

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

Training loop

```python
lr = 0.1

for epoch in range(1000):
    # forward
    z = X @ W + b
    y_hat = sigmoid(z)

    # loss (MSE)
    loss = np.mean((y - y_hat) ** 2)

    # backward
    dZ = (y_hat - y) * (y_hat * (1 - y_hat))
    dW = X.T @ dZ
    db = np.sum(dZ)

    # update
    W -= lr * dW
    b -= lr * db

    if epoch % 100 == 0:
        print(loss)
```

Test

```python
print(sigmoid(X @ W + b))
```
