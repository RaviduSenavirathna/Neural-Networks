AND gate implimentation

```python
import numpy as np

# Define the dataset for the AND problem
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [0],
    [0],
    [1]
])
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Initialize weights and bias
Weights = np.random.randn(2, 1)
bias = np.zeros((1, 1))
```

```python
# Sigmoid activation function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
# Training loop
learning_rate = 0.1

for epoch in range(1000):
    # forward
    z = X @ Weights + bias
    y_hat = sigmoid(z)

    # loss (MSE)
    loss = np.mean((y - y_hat) ** 2)

    # backward
    dZ = (y_hat - y) * (y_hat * (1 - y_hat))
    dW = X.T @ dZ
    db = np.sum(dZ)

    # update
    Weights -= learning_rate * dW
    bias -= learning_rate * db

    if epoch % 100 == 0:
        print(loss)
```

```python
# After training, print the final predictions
print(sigmoid(X @ Weights + bias))
```
